```python
# Connect to the provided database
import sqlite3

connection = sqlite3.connect("file:my_data.db?mode=ro", uri=True)
cursor = connection.cursor()
print("Database connected.")
```

```python
# 1 Run the SQL query
cursor.execute("""
SELECT sale_date, product, price
FROM sales;
""")

# Display the column names and results
for column in cursor.description:
    print(column[0], end=" | ")
print()

for row in cursor.fetchall():
    print(row)
```

```python
#2
cursor.execute("""
SELECT *
FROM employees
WHERE city = 'Chicago' AND years_experience > 10;
""")

# Display the column names and results
for column in cursor.description:
    print(column[0], end=" | ")
print()

for row in cursor.fetchall():
    print(row)
```

```python
# 3
cursor.execute("""
SELECT *
FROM listings
ORDER BY sqft DESC
LIMIT 3;
""")

# Display the column names and results
for column in cursor.description:
    print(column[0], end=" | ")
print()

for row in cursor.fetchall():
    print(row)
```

```python
# 4
cursor.execute("""
SELECT week, SUM(impressions) AS total_impressions
FROM campaigns
GROUP BY week
ORDER BY week;
""")

# Display the column names and results
for column in cursor.description:
    print(column[0], end=" | ")
print()

for row in cursor.fetchall():
    print(row)
```

```python
# 5
cursor.execute("""
SELECT order_date, ROUND(SUM(order_total), 2) AS total_sales
FROM orders
GROUP BY order_date
HAVING SUM(order_total) > 400
ORDER BY order_date;
""")

# Display the column names and results
for column in cursor.description:
    print(column[0], end=" | ")
print()

for row in cursor.fetchall():
    print(row)
```

```python
# 6
cursor.execute("""
SELECT doctor, ROUND(AVG(patient_age), 2) AS average_patient_age,
       MAX(wait_minutes) AS longest_wait_minutes
FROM visits
GROUP BY doctor
ORDER BY doctor;
""")

# Display the column names and results
for column in cursor.description:
    print(column[0], end=" | ")
print()

for row in cursor.fetchall():
    print(row)
```

```python
#7
cursor.execute("""
SELECT customers.branch, ROUND(SUM(transactions.amount), 2) AS total_transaction_amount
FROM customers
JOIN transactions ON customers.customer_id = transactions.customer_id
GROUP BY customers.branch
ORDER BY customers.branch;
""")

# Display the column names and results
for column in cursor.description:
    print(column[0], end=" | ")
print()

for row in cursor.fetchall():
    print(row)
```
